# Week 2 Assignment: Temperature Conversion and Control Structures Program

```python
# ==========================================
# Week 2 Assignment: Temperature Converter
# ==========================================

# 1. Define a function to perform the temperature conversion from Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    """Converts a given temperature in Celsius to Fahrenheit."""
    return (celsius * 9/5) + 32

# Set a specified threshold temperature in Celsius
THRESHOLD_C = 25.0

print("=== Part 1: Interactive User Input Conversion ===")
try:
    # Ask the user to enter a temperature in °C
    user_input = float(input("Enter temperature in °C: "))

    # Convert the entered temperature using the function
    converted_f = celsius_to_fahrenheit(user_input)
    print(f"{user_input}°C is equal to {converted_f:.2f}°F.")

    # Report whether the temperature is above or below the threshold
    if user_input > THRESHOLD_C:
        print(f"Report: The entered temperature is ABOVE the threshold ({THRESHOLD_C}°C).")
    elif user_input < THRESHOLD_C:
        print(f"Report: The entered temperature is BELOW the threshold ({THRESHOLD_C}°C).")
    else:
        print(f"Report: The entered temperature is EXACTLY at the threshold ({THRESHOLD_C}°C).")

except ValueError:
    print("Invalid input! Please enter a numeric value for the temperature.")


print("\n=== Part 2: Loop to Convert Several Temperatures ===")
# Define a list of several temperatures to test
temperatures_c = [0, 10, 18, 25, 32, 38, 45]

print(f"{'Celsius (°C)':<15} | {'Fahrenheit (°F)':<15} | {'Status (vs 25°C Threshold)'}")
print("-" * 65)

# Use a loop to iterate through and convert several temperatures
for temp_c in temperatures_c:
    temp_f = celsius_to_fahrenheit(temp_c)

    # Check threshold status
    if temp_c > THRESHOLD_C:
        status = "Above threshold"
    elif temp_c < THRESHOLD_C:
        status = "Below threshold"
    else:
        status = "At threshold"

    # Display the results neatly
    print(f"{temp_c:<15} | {temp_f:<15.2f} | {status}")
```
