```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing

# 1. Load the dataset from Week 3
housing = fetch_california_housing(as_frame=True)
df = housing.frame

# 2. Select 4 numerical variables relevant to the project
selected_vars = ['MedInc', 'AveRooms', 'HouseAge', 'MedHouseVal']
subset_df = df[selected_vars]

# 3. Calculate mean, median, variance, standard deviation, and range for each variable
stats = pd.DataFrame({
    'Mean': subset_df.mean(),
    'Median': subset_df.median(),
    'Variance': subset_df.var(),
    'Standard Deviation': subset_df.std(),
    'Range': subset_df.max() - subset_df.min()
})

print("=== Descriptive Statistics ===")
display(stats)

# 4. Select two variables (MedInc and MedHouseVal) to calculate covariance and correlation
cov_matrix = subset_df[['MedInc', 'MedHouseVal']].cov()
corr_matrix = subset_df[['MedInc', 'MedHouseVal']].corr()

print("\n=== Covariance Matrix ===")
display(cov_matrix)

print("\n=== Correlation Matrix ===")
display(corr_matrix)
```

## Statistical Insights and Explanations

* **What the statistics tell us about the data:**
  - The **mean** and **median** provide the baseline central tendency, showing typical values for neighborhood income and home values.
  - The **variance** and **standard deviation** measure how spread out or dispersed the data points are from the mean, showing the diversity of property values and income levels across different regions.
  - The **range** outlines the full span from the minimum to the maximum observed values in the dataset.

* **What the correlation tells us about the relationship:**
  - The correlation coefficient between Median Income (`MedInc`) and Median House Value (`MedHouseVal`) is strong and positive (approximately 0.69). This indicates that blocks with higher median incomes consistently tend to have higher median property values.

* **Does correlation necessarily mean that one variable causes the other?**
  - No, correlation does not imply causation. While higher income is strongly associated with higher housing values, it does not mean that increasing a neighborhood's income directly causes its home values to rise in absolute isolation; other underlying factors (such as neighborhood infrastructure, property size, and market demand) also heavily influence home prices.
